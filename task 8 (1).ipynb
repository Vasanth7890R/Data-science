```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
penguins = sns.load_dataset("penguins")
penguins.head()
```

```python
print(penguins.shape)
print(penguins.columns)
print(penguins.dtypes)
print(penguins.isnull().sum())
```

```python
for col in penguins.select_dtypes(include="number").columns:
    penguins[col] = penguins[col].fillna(penguins[col].median())
```

```python
print(penguins.isnull().sum())
```

```python
for col in penguins.select_dtypes(include="number").columns:
    penguins[col] = penguins[col].fillna(penguins[col].median())
```

```python
print(penguins.isnull().sum())
```

```python
print(penguins["species"].value_counts())
```

```python
print(penguins["species"].value_counts(normalize=True) * 100)
```

```python
print(penguins["species"].mode()[0])
```

```python
sns.countplot(data=penguins, x="species")
plt.title("Species Count")
plt.show()
```

```python
penguins["island"].value_counts().plot.pie(autopct="%1.1f%%")
plt.title("Island Distribution")
plt.show()
```

```python
penguins["island"].value_counts().plot(kind="bar")
plt.title("Island Distribution")
plt.xlabel("Island")
plt.ylabel("Count")
plt.show()
```

```python
sns.countplot(data=penguins, x="species", hue="sex")
plt.title("Species Comparison")
plt.show()
```

```python
encoded = pd.get_dummies(penguins, columns=["species", "island", "sex"], dtype=int)
print(encoded.head())
```
