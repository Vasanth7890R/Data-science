```python
for col in df.select_dtypes(include='number').columns:
    df[col] = df[col].fillna(df[col].median())

for col in df.select_dtypes(include=['object', 'str']).columns:
    df[col] = df[col].fillna(df[col].mode()[0])

df.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder

df_encoded = df.copy()

for col in df_encoded.select_dtypes(include='str').columns:
    le = LabelEncoder()
    df_encoded[col] = le.fit_transform(df_encoded[col])

df_encoded.head()
```

```python
sns.scatterplot(data=df, x='lead_time', y='adr', hue='hotel')

plt.title('Lead Time vs Average Daily Rate')
plt.xlabel('Lead Time')
plt.ylabel('ADR')
plt.tight_layout()
plt.show()
```

```python
sns.violinplot(data=df, x='hotel', y='adr')

plt.title('ADR Distribution by Hotel Type')
plt.xlabel('Hotel Type')
plt.ylabel('Average Daily Rate')
plt.tight_layout()
plt.show()
```

```python
skewness = df['adr'].skew()

sns.boxplot(data=df, x='hotel', y='adr')

plt.title(f'ADR Box Plot | Skewness: {skewness:.2f}')
plt.xlabel('Hotel Type')
plt.ylabel('Average Daily Rate')

plt.tight_layout()
plt.show()

print("Skewness of ADR:", round(skewness, 2))
```
